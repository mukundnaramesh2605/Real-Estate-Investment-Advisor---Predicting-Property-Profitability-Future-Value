# Encode the columns in the datasets

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder,MultiLabelBinarizer
lable = LabelEncoder()
multi_label = MultiLabelBinarizer()
```

```python
df = pd.read_csv(r'Datasets/india_housing_prices_with_target_columns.csv')
df.head()
```

### One Hot Encoding the City column


```python
df['City'] = df['City'].str.replace(' ', '_')
df = pd.get_dummies(df,columns=['City'],prefix='City',dtype=int)

df.head()
```

### One Hot Encoding the Property Type column

```python
df['Property_Type'] = df['Property_Type'].str.replace(' ', '_')
df = pd.get_dummies(df,columns=['Property_Type'],prefix='Property_Type',dtype=int)
df.head()
```

### One Hot Encoding the Facing column

```python
df = pd.get_dummies(df,columns=['Facing'],prefix='Facing',dtype=int)
df.head()
```

### One Hot Encoding the Furnished Status column

```python
df['Furnished_Status'] = df['Furnished_Status'].str.replace('-', ' ').str.title().str.replace(' ', '_')
df = pd.get_dummies(df,columns=['Furnished_Status'],prefix='Furnished_Status',dtype=int)
df.head()
```

### One Hot Encoding the Owner Type column

```python
df = pd.get_dummies(df,columns=['Owner_Type'],prefix='Owner_Type',dtype=int)
df.head()
```

### Label Encoding the Availability Status column

```python
df['Availability_Status'] = lable.fit_transform(df['Availability_Status'])
class_mapping = dict(zip(lable.classes_, range(len(lable.classes_))))
print("Class Mapping:", class_mapping)
```

### Label Encoding the Parking Space column

```python
df['Parking_Space'] = lable.fit_transform(df['Parking_Space'])
class_mapping = dict(zip(lable.classes_, range(len(lable.classes_))))
print("Class Mapping:", class_mapping)
```

### Label Encoding the Security column

```python
df['Security'] = lable.fit_transform(df['Security'])
class_mapping = dict(zip(lable.classes_, range(len(lable.classes_))))
print("Class Mapping:", class_mapping)
```

### Ordinal Encoding the Public Transport Accessibility column

```python
unique_condition = ['Low','Medium','High']
conditions_mapping = {k:i for i,k in enumerate(unique_condition)}
df['Public_Transport_Accessibility'] = df['Public_Transport_Accessibility'].map(conditions_mapping)
```

```python
df.to_csv('Datasets/india_housing_prices_with_target_columns_encoded.csv',index=False)
```
