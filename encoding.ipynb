# Encode the columns in the datasets

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder,MultiLabelBinarizer
```

```python
df = pd.read_csv(r'Datasets/india_housing_prices_with_target_columns.csv')
df.head()
```

### One Hot Encoding the City column


```python
df = pd.get_dummies(df,columns=['City'],prefix='city',dtype=int)
df.head()
```

### One Hot Encoding the Property Type column

```python
df = pd.get_dummies(df,columns=['Property_Type'],prefix='property_type',dtype=int)
df.head()
```
